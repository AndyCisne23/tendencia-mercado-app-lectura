### Encontrar número de libros publicados después del 1 de enero de 2000

### Encontrar el número de reseñas de usuarios y la calificación promedio para cada libro

```python
# Se junta la tabla 'books' con 'ratings' y con 'reviews' para poder hacer la consulta
# Se cuenta el número de reseñas de los usurios y la califiación promedio para cada libro

query_reviews_ratings = '''
SELECT
    b.book_id,
    b.title,
    COUNT(rv.review_id) AS num_reviews,
    AVG(rt.rating) AS avg_rating
FROM
    books b
LEFT JOIN
    reviews rv ON b.book_id = rv.book_id
LEFT JOIN
    ratings rt ON b.book_id = rt.book_id
GROUP BY
    b.book_id, b.title
ORDER BY
    avg_rating DESC
'''

books_reviews_ratings = pd.io.sql.read_sql(query_reviews_ratings, con = engine)

print(f'Tabla con la cantidad de reseñas por libro [num_reviews] y calificación\npromedio [avg_rating]\n')
display(books_reviews_ratings)
```

### Comentarios:

Con esta consulta que está acomodada de forma descendente según el promedio de calificación que tiene cada libro se pueden observar algunas cosas.

Por ejemplo, los libros que tienen una calificaciín más alta no tiene un gran número de reseñas, esto puede indicar que podría haber suarios que califican el titulo sin dejar ningún comentario al respecto o que no muchos conocen el libro/autor, sin embargio a quienes lo conocen y lo han leido les ha encantado y por eso puntuan con la mayor calificación.

Aquellos libros que tienen muchas reseñas y una calificación promedio baja probablemente indican que son libros populares pero no han logrado satisfacer las expectaitvas que tiene el público en generla respecto a ellos.

### Identificar editorial que ha publicado mayor número de libros con más de 50 páginas

```python
# Se juntan las tablas de 'books' y 'publishers' para poder hacer la consulta
# Se identifican las editoriales que han publicado el mayor número de libros con más de 50 páginas

query_books_publisher = '''
SELECT
    p.publisher,
    COUNT(b.title) AS publisher_count
FROM
    publishers p
JOIN
    books b ON p.publisher_id = b.publisher_id
WHERE b.num_pages > 50
GROUP BY
    p.publisher
ORDER BY
    publisher_count DESC
'''

books_publishers = pd.io.sql.read_sql(query_books_publisher, con = engine)
publisher_name = books_publishers.iloc[0,0]
publisher_count = books_publishers.iloc[0,1]

print(f'La editorial que ha publicado mayor número de libros con más de 50 páginas es: {publisher_name} con {publisher_count} libros publicados')
print(f'\nTabla con datos:\n')
display(books_publishers)
```

### Comentarios:

Para quienes son ávidos lectores, se sabe que le editorial 'Penguin' es una de las más grandes de todas, por lo que no es de sorprender que se encuentre en el lugar en el que está, incluso se puede apreciar en la previsualización de la tabla que hay otra categoria llamada 'Penguin Classics' la cual, aunque esta separa en el nombre se sabe que es también parte de la familia 'Penguin' por lo que hace que el número se eleve todavía más si es que en algún momento del análsis se tuvieran que sumar.

# Identificar al autor que tiene la más alta calificación promedio del libro

Se consideran sólo aquellos libros que tienen al menos 50 reseñas.

```python
# Se juntan las tablas 'books', 'ratings' y 'authors' para la consulta
# Se identifica al autor que tiene la más alta calificación promedio del libro
# y se toma en cuenta sólo aquellos libros que tienen al menos 50 reseñas

query_author_rating = '''
SELECT
    a.author,
    COUNT(rt.rating) AS rating_count,
    AVG(rt.rating) AS rating_avg
FROM
    books b
JOIN
    ratings rt ON rt.book_id = b.book_id
JOIN
    authors a ON a.author_id = b.author_id
GROUP BY a.author
HAVING COUNT(rt.rating) >= 50
ORDER BY rating_avg DESC
'''

author_rating = pd.io.sql.read_sql(query_author_rating, con = engine)
author_name = author_rating.iloc[0,0]
author_rating_avg = author_rating.iloc[0,2]

print(f'El autor que tiene la más alta calificación promedio del libro es:\n{author_name} con una califciación promedio de {author_rating_avg}')
print(f'\nTabla con datos:\n')
display(author_rating)
```

### Comentarios:

Con la consulta anterior se puede identificar de manera muy sencilla, cual es el autor que tiene un mejor promedio en calificación respecto a sus libros con más de 50 reseñas en la aplicación.

Se observa que no siempre los autores con mayor cantidad de reseñas son quienes tienen una puntuación mucho mayor. Con frecuencia los usuarios tienden a dejar más malos comentarios que buenos, por lo tanto puede ser que una gran cantidad de reseñas no siempre signifique que un libro es bueno o muy popular.

Muchos autores y libros tienen una base fe fans que son muy leales y por ende califican alto los titulos, por ejemplo el caso de J.K. Rowling con 'Harry Potter', quien a pesar de ser una autora con tanto éxito y que sus libros han marcado a toda una generación no tiene la puntuación más alta a pesar de encabezar el número de reseñas.

Además es importante tomar en cuenta, como ya se mencionó con anterioridad que muchas veces los usuarios pueden calificar el titulo, ya se de forma positiva o negativa, pero no dejan reseñas o coemntarios del mismom, por lo que guiarse sólo en la cantidad de comentarios para tomar cualquier decisión no es lo más recomendable.

```python
# Se juntan las tablas de 'ratings' y 'review' para hacer la consulta
# Se encuentra el número promedio de reseñas de texto entre usuarios que calificaron más de 50 libros


query_ratings_review = '''
SELECT
    AVG(total_reviews)
FROM(
SELECT
    rt.username,
    rv.total_reviews
FROM (
    SELECT
        username,
    COUNT(*)
        AS total_ratings
    FROM
        ratings
    GROUP BY
        username
    HAVING
        COUNT(*) > 50
) rt
INNER JOIN (
    SELECT
        username,
    COUNT(*)
        AS total_reviews
    FROM
        reviews
    GROUP BY
        username
) rv ON rt.username = rv.username
) AS subquery
'''


rating_review_avg = pd.io.sql.read_sql(query_ratings_review, con = engine)
text_rating_avg = rating_review_avg.iloc[0, 0]

print(f'El promedio de número de reseñas de texto entre los usuarios que calificaron\nmás de 50 libros es de: {text_rating_avg} reseñas')
```

### Comentarios:

Siguiendo con los comentarios anteriores, aquí se puede observar que la cantidad promedio de reseñas que dejan los usuarios que han calificado más de 50 libros es de 24.

Esto quiere decir que los usuarios sólo dejan una reseña o comentario a 1/4 parte de los libros que leen y que califican.
